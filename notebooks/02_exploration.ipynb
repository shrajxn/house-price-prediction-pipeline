```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["SalePrice"], bins=30)
plt.title("SalePrice Distribution")
plt.xlabel("SalePrice")
plt.ylabel("Frequency")
plt.show()
```

```python
correlation_matrix = df.corr(numeric_only=True)
```
