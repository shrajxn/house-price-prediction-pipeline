```python
import pandas as pd
import sys
from pathlib import Path

sys.path.append(str(Path().resolve().parent))

from src.data.load_data import load_csv_data
```

```python
df=load_csv_data("../data/processed/train_cleaned.csv")
```

```python
df["HouseAge"] = (
    df["YrSold"] -
    df["YearBuilt"]
)
```

```python
df[
    ["YearBuilt",
     "YrSold",
     "HouseAge"]
].head()
```

```python
df["TotalBathrooms"] = (
    df["FullBath"]
    + 0.5 * df["HalfBath"]
    + df["BsmtFullBath"]
    + 0.5 * df["BsmtHalfBath"]
)
```

```python
df[
    [
        "FullBath",
        "HalfBath",
        "BsmtFullBath",
        "BsmtHalfBath",
        "TotalBathrooms"
    ]
].head()
```

```python
df["TotalSquareFeet"] = (
    df["TotalBsmtSF"]
    + df["1stFlrSF"]
    + df["2ndFlrSF"]
)
```

```python
df[
    [
        "TotalBsmtSF",
        "1stFlrSF",
        "2ndFlrSF",
        "TotalSquareFeet"
    ]
].head()
```
